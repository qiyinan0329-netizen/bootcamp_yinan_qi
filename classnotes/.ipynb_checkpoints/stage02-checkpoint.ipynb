```python
def sqrt(x):
    x = float(abs(x))
    r0 = x
    r1 = x/2
    while abs(r0-r1)>10**-12
    r0 = r1
    r1 = (r0+x/r0)/2
    
return r1
```

```python
def mysqrt(x):
    x = float(abs(x))
    r0 = x
    r1 = x/2
    while abs(r0-r1)>10e-7:
        r0 = r1
        r1 = (r0+x/r0)/2
        print("r1:",r1)
    return r1
```

```python
def auth(quote,**speaker_info):
    print(speaker_info)
    print(type(speaker_info))
    
    print(">",quote)
    print("-"*(len(quote)+2))
    for k,v in speaker_info.items():
        print(k,v,sep=": ")
```

```python
class rect:
    def __init__(self, width=0, height=0):
        self.x = 0
        self.y = 0
        self.width = width
        self.height = height
    
    def setxy(self, dx, dy):
        self.x += dx
        self.y += dy
        
    def grow(self, dw, dh):
        self.width += dw
        self.height += dh
    
    def area(self):
        return self.width*self.height
    
    def info(self):
        outs ="({0},{1} width: {2},height: {3})"
        print(outs.format(self.x, self.y, self.width, self.height))
        
    def pointin(self, p):
        return self.x<p.x and self.x+self.width>p.x
        
    def __repr__(self):
        outs = "({0},{1} width:{2}, height:{3})"
        return(outs.format(self.x, self.y, self.width, self.height))
```

```python
r4 = rect(1,2)#调用__init__
r4.info()
```

```python
r4.setxy(10,20)
print(r4)#调用__repr__
```

```python
r4.grow(5,5)
r4.info()
```

```python
print(r4,area(r4))
```

```python
class point:
    def __init__(self,x=0,y=0):
        self.x = x
        self.y = y
        
    def move(self, dx, dy):
        self.x += dx
        self.y += dy
    
    def info(self):
        print("横坐标:{0} 纵坐标:{1}".format(self.x, self.y))
        
    def setxy(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return "("+str(self.x)+", "+str(self.y)+")"
    
    def distance(self, p):
        return((self.x-p.x)**2+(self.y-p.y)**2)**0.5
```

```python
p1 = point(1,2)#调用__init__
print(p1.x, p1.y)
```

```python
p2 = point(10,20)
print(p2.x, p2.y)
```

```python
p1.move(2,3)
print(p1.x, p1.y)
```

```python
p2.info()
```

```python
p2.setxy(15,25)
p2.info()
```

```python
print(p1)
```

```python
p1.distance(p2)
```
